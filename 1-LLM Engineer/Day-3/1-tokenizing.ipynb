```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4o")
input_text = """
Akalanka Sakalasooriya is a professional AI and Machine Learning Engineer based in the Western Province of Sri Lanka. He specializes in developing efficient machine learning models, natural language processing (NLP) applications, and backend systems
"""
tokens = encoding.encode(input_text)
print(tokens)
print("length of tokens: ", len(tokens))
```
